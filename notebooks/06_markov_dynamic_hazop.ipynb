# Markov-modell és Dynamic HAZOP

Állapotvalószínűségek és a maintenance rate hatása egy aggregált TK-101 modellben.

```python
import numpy as np
import matplotlib.pyplot as plt
from safetycourse.markov import transient_probabilities
```

```python
def Q_matrix(mu_maint=0.08):
    ls, lc, lr, lo = 0.002, 0.004, 0.20, 0.01
    return np.array([
      [-(ls+lc), ls, lc, 0.0],
      [mu_maint, -(mu_maint+lc), lc, 0.0],
      [lr, 0.0, -(lr+lo), lo],
      [0.0,0.0,0.0,0.0]
    ])

times=np.linspace(0,250,251)
p0=np.array([1.,0.,0.,0.])
P=transient_probabilities(Q_matrix(),p0,times)
P[-1]
```

```python
for i,name in enumerate(['Normal','Sensor degraded','Challenge','Overflow']):
    plt.plot(times,P[:,i],label=name)
plt.xlabel('time [h]'); plt.ylabel('state probability'); plt.legend(); plt.show()
```

## Maintenance sensitivity

```python
for mu in [0.02,0.08,0.20]:
    p=transient_probabilities(Q_matrix(mu),p0,[100])[0]
    print('mu=',mu,'P(overflow at 100h)=',round(p[3],6))
```

**Dynamic HAZOP kérdés:** mely deviationoknál változtatja meg a következményt a válaszidő/sorrend?
