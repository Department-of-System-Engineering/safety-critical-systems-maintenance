# Fault Tree Analysis – TK-101

Automatikusan generált fault tree, minimal cut set és Birnbaum importance.

```python
from pathlib import Path
from safetycourse.model_io import load_yaml, generate_fault_tree
from safetycourse.fta import top_event_probability, minimal_cut_sets, birnbaum_importance
from safetycourse.plotting import draw_fault_tree
import matplotlib.pyplot as plt
```

```python
model = load_yaml('../data/pundit_case/system_model.yaml')
tree = generate_fault_tree(model, 'OVERFLOW')
print('P(OVERFLOW)=', top_event_probability(tree))
print('Minimal cut sets:')
for c in minimal_cut_sets(tree): print(sorted(c))
```

```python
importance = birnbaum_importance(tree)
sorted(importance.items(), key=lambda x:x[1], reverse=True)
```

```python
ax = draw_fault_tree(tree)
plt.show()
```

**Feltételezés:** a közvetlen kapubemenetek függetlensége. Shared basic event/common cause esetén külön modell szükséges.
