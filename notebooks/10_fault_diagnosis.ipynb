# Fault diagnosis – LT-101

Residual, EWMA és threshold detector szintetikus ground truth mellett.

```python
import pandas as pd
import matplotlib.pyplot as plt
from safetycourse.diagnosis import residual, ewma, threshold_detector, binary_metrics
```

```python
df=pd.read_csv('../data/pundit_case/sensor_data.csv')
df['residual']=residual(df.lt101_measured_pct,df.true_level_pct)
df['ewma']=ewma(df.residual,alpha=0.12)
df.head()
```

```python
for gamma in [2,4,6]:
    pred=threshold_detector(df.ewma,gamma,'absolute')
    print(gamma,binary_metrics(df.fault_active,pred))
```

```python
plt.plot(df.time_min,df.residual,label='residual')
plt.plot(df.time_min,df.ewma,label='EWMA')
plt.xlabel('time [min]'); plt.ylabel('level residual [%]'); plt.legend(); plt.show()
```

`true_level_pct` oktatási ground truth. Valós üzemben referencia vagy modellbecslés szükséges.
