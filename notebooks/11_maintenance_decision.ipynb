# Risk-based maintenance decision

Expected-cost scenario comparison és érzékenység a consequence cost-ra.

```python
from safetycourse.maintenance import compare_maintenance_options
```

```python
def decide(consequence=50_000_000, wait_p=0.02):
    return compare_maintenance_options([
      {'name':'maintain_now','maintenance_cost':250_000,'production_loss':800_000,'hazard_probability':0.001,'consequence_cost':consequence},
      {'name':'wait_for_window','maintenance_cost':250_000,'production_loss':100_000,'hazard_probability':wait_p,'consequence_cost':consequence},
    ])
decide()
```

```python
for c in [5e6,10e6,25e6,50e6,100e6]:
    ranked=decide(consequence=c)
    print(int(c), ranked[0]['name'], round(ranked[0]['expected_cost']))
```

Az expected cost nem safety acceptance criterion; a példában döntési proxy.
