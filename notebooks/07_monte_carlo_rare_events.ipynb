# Monte Carlo és ritka események

Crude Monte Carlo, mintaszám-hatás és splitting/RESTART-intuíció.

```python
from safetycourse.simulation import bernoulli_system_monte_carlo, random_walk_exceedance, multilevel_splitting_random_walk
```

```python
p={'challenge':0.08,'lshh':0.01,'xv':0.005,'sensor':0.02,'operator':0.04}
def logic(x):
    return x['challenge'] & (x['lshh']|x['xv']) & (x['sensor']|x['operator'])
for n in [1_000,10_000,100_000,1_000_000]:
    print(n, bernoulli_system_monte_carlo(p, logic, n=n, seed=11))
```
