# Bayesian Network – diagnosztikai inference

Prior → evidence → posterior egy egyszerű LT-101 hibamodellen.

```python
from safetycourse.bayes import BinaryNode, BinaryBayesNet
```

```python
net = BinaryBayesNet([
 BinaryNode('LT101Fault', [], {():0.02}),
 BinaryNode('SensorMismatch', ['LT101Fault'], {(False,):0.05,(True,):0.90}),
])
print('prior', net.query('LT101Fault')[True])
print('posterior | mismatch', net.query('LT101Fault', {'SensorMismatch':True})[True])
print('posterior | no mismatch', net.query('LT101Fault', {'SensorMismatch':False})[True])
```

## Sensitivity to false-positive rate

```python
for fp in [0.01,0.05,0.10,0.20]:
    n = BinaryBayesNet([BinaryNode('F', [], {():0.02}), BinaryNode('E',['F'],{(False,):fp,(True,):0.90})])
    print(fp, round(n.query('F', {'E':True})[True],4))
```
